# Exploring Professional Tennis Career Matches

This notebook demonstrates the `tennis_career` Python package developed for STAT 159 HW3. The package retrieves and processes professional tennis match data from Tennis Abstract and generates NYT-style scatter plots comparing players' ages with their opponents' ages.

We demonstrate the package using Serena Williams and Rafael Nadal.

```python
import matplotlib.pyplot as plt

from tennis_career import get_player_matches, plot_age_difference
```

## Serena Williams

Retrieve Serena Williams's career match data and examine the processed DataFrame. Each row represents a match, including the result, opponent, player's age, opponent's age, and age difference.

```python
serena_df = get_player_matches("Serena Williams", "female")

print("Number of matches:", len(serena_df))
display(serena_df.head())
```

```python
fig, ax = plot_age_difference(
    serena_df,
    player_name="Serena Williams"
)

plt.show()
```

## Rafael Nadal

To demonstrate that the package is reusable, we apply the same functions to Rafael Nadal's career matches.

```python
nadal_df = get_player_matches("Rafael Nadal", "male")

print("Number of matches:", len(nadal_df))
display(nadal_df.head())
```

```python
fig, ax = plot_age_difference(
    nadal_df,
    player_name="Rafael Nadal"
)

plt.show()
```

## Conclusion

The `tennis_career` package provides a reusable workflow for retrieving, cleaning, and visualizing professional tennis match data. The scatter plots show how each player's age relative to their opponents changes throughout their career, with colors distinguishing wins from losses.

The notebook demonstrates both public functions without requiring users to interact directly with the package's private helper functions.

```python
print("Number of matches:", len(nadal_df))
print("Earliest match:", nadal_df["match_date"].min())
print("Latest match:", nadal_df["match_date"].max())
print("Minimum age:", nadal_df["player_age"].min())
print("Maximum age:", nadal_df["player_age"].max())
```
